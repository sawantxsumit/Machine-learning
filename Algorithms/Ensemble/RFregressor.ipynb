```python
# Check for null values
df.isnull().sum()
```

```python
# check for missing values
features_with_na= [feature for feature in df.columns if df[feature].isnull().sum() >=1 ]
features_with_na
```

```python
# percentage of missing values
for feature in features_with_na:
    print(feature , np.round(df[feature].isnull().mean()*100,2), '% missing values')
```

```python
df.head()
```

```python
df['fuel_type'].value_counts()
```

```python
df['fuel_type'] = df['fuel_type'].replace('not supported', '–')
```

```python
df['fuel_type'].value_counts()
```

```python
df['fuel_type'].fillna('Gasoline', inplace=True)
```

```python
df['accident'].value_counts()
```

```python
df['accident'].fillna('None reported', inplace=True)
```

```python
''' 
Clean Title: Evaluate the availability of a clean title, which can impact the vehicle's resale value and legal status.
'''
df['clean_title'].value_counts()
```

```python
df['clean_title'].fillna('No', inplace=True)
```

```python
df.isnull().sum()
```

```python
df.dtypes
```

```python
df['milage'] = df['milage'].astype(str)  # convert to string
df['milage'] = df['milage'].str.replace('mi.', '', regex=False)  # remove 'mi.'
df['milage'] = df['milage'].str.replace(',', '')  # remove commas
df['milage'] = df['milage'].str.strip()  # remove any leading/trailing whitespace
df['milage'] = df['milage'].astype(int)  # convert to integer
```

```python
df['milage']
```

```python
df.head()
```

```python
df.dtypes
```

```python
df['engine'].unique()
```

```python
df.drop('engine', axis=1 , inplace=True)
```

```python
df['transmission'].value_counts()
```

```python
# changing data typeof price column
df['price'] = df['price'].astype(str)  # convert to string
df['price'] = df['price'].str.replace('$', '', regex=False)  # remove 'mi.'
df['price'] = df['price'].str.replace(',', '')  # remove commas
df['price'] = df['price'].str.strip()  # remove any leading/trailing whitespace
df['price'] = df['price'].astype(int)  # convert to integer
```

```python
df.dtypes
```

```python
# getting numerical features
num_features=[feature for feature in df.columns if df[feature].dtype != 'O']
print('Num of Numerical features are ', len(num_features))
```

```python
#categorical features
cat_features=[feature for feature in df.columns if feature not in num_features]
print('Num of categorical features are ', len(cat_features))
```

```python
# discrete features
dis_features= [feature for feature in df.columns if len(df[feature].unique()) <=25]
print('Num of discrete features are ', len(dis_features))
```

```python
x=df.drop('price', axis=1)
y=df['price']
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.linear_model import LogisticRegression
```

```python
# Column transformer
cat_features= x.select_dtypes(include='object').columns

num_features= x.select_dtypes(exclude='object').columns
from sklearn.preprocessing import OneHotEncoder , StandardScaler
from sklearn.compose import ColumnTransformer
num_transformer= StandardScaler()
onehot= OneHotEncoder(drop='first', handle_unknown='ignore')

```

```python
preprocessor=ColumnTransformer(
    [
        ('One hot' , onehot , cat_features),
        ('Standard' , num_transformer , num_features)
    ]
)
```

```python
# Train test split
from sklearn.model_selection import train_test_split
x_train , x_test , y_train , y_test = train_test_split(x,y, test_size=0.3 , random_state=22)
# Better approach:
from sklearn.pipeline import Pipeline 
from sklearn.preprocessing import StandardScaler 
# Create pipeline
pipeline = Pipeline([ ('preprocessor', preprocessor), ('model', RandomForestRegressor()) ]) 
# Fit on training data only 
pipeline.fit(x_train, y_train)
y_pred = pipeline.predict(x_test)
```

```python
x_train=preprocessor.fit_transform(x_train)
x_test= preprocessor.transform(x_test)
```

```python
pd.DataFrame(x_train).head()
```

```python
x_test
```

```python

from sklearn.metrics import mean_absolute_error ,mean_squared_error , r2_score , root_mean_squared_error

```

```python
models={
    'Random forest':RandomForestRegressor(),
    'Decision Tree':DecisionTreeRegressor(),
    'Logistic regression':LogisticRegression()
}
for i in range(len(list(models))):
    model=list(models.values())[i]
    model.fit(x_train , y_train)
    
    y_train_pred= model.predict(x_train)
    y_test_pred= model.predict(x_test)
    
    #training set performance
    model_train_rsme=root_mean_squared_error(y_train , y_train_pred)
    model_train_mae= mean_absolute_error(y_train , y_train_pred )
    model_train_mse= mean_squared_error(y_train , y_train_pred)
    model_train_r2score= r2_score(y_train , y_train_pred)
    
    #test set performance
    model_test_rsme=root_mean_squared_error(y_test , y_test_pred)
    model_test_mae= mean_absolute_error(y_test , y_test_pred )
    model_test_mse= mean_squared_error(y_test , y_test_pred)
    model_test_r2score= r2_score(y_test , y_test_pred)
    
    print(list(models.keys())[i])
    
    print('Model performance for training set')
    print('- Mean squared error :{:.4f}'.format(model_train_mse))
    print('- Mean absolute error: {:.4f}'.format(model_train_mae))
    print('- Root mean squared error : {:.4f}'.format(model_train_rsme))
    print('- R2_score : {:.4f}'.format(model_train_r2score))
    
    print('='*35)
    
    print('Model performance for test set')
    print('- Mean squared error :{:.4f}'.format(model_train_mse))
    print('- Mean absolute error: {:.4f}'.format(model_train_mae))
    print('- Root mean squared error : {:.4f}'.format(model_train_rsme))
    print('- R2_score : {:.4f}'.format(model_train_r2score))
    
    print('='*35)
    print()
```

```python
# Hyperparameter tuning 

from sklearn.model_selection import GridSearchCV
rf=RandomForestRegressor(random_state=22)
rf_param = {
    'n_estimators': [100, 200],
    'max_depth': [None, 10, 20],
    'min_samples_split': [2, 5],
    'min_samples_leaf': [1, 2],
    'max_features': ['sqrt', 'log2']
}

gdr= GridSearchCV(estimator=rf , param_grid=rf_param , cv=5 , n_jobs=-1 , verbose=2, scoring='r2')
```

```python
gdr.fit(x_train , y_train)
```

```python
# Best paramters 
print('Best parameters :', gdr.best_params_)
print('Best r2 score from cv :' , gdr.best_score_)
```

```python
#Evaluate on test set
best_model= gdr.best_estimator_
y_pred= best_model.predict(x_test)
```

```python
print("Test R2 score:", r2_score(y_test, y_pred))
print("Test MAE:", mean_absolute_error(y_test, y_pred))
print("Test RMSE:", root_mean_squared_error(y_test, y_pred))
```
